```python
import flock
```

## 1. Ping

2 workers. Rank 0 sends `"ping"` to rank 1. Rank 1 receives it and returns the message. Other ranks return `None`.


Expected: `[None, "ping"]`

```python
@flock.distribute(workers=2, seed=2)
async def ping():
    if flock.rank() == 0:
        await flock.isend(1, "ping")
        return None
    else:
        flock.recv(0)


ping()
```

## 2. Sum

4 workers. Each rank contributes its own `rank` to an `all_reduce` with `"sum"`. Return the result.

Expected: `[6, 6, 6, 6]`

```python
@flock.distribute(workers=4)
async def sum_ranks():
    return await flock.all_reduce(flock.rank(), "sum")
    


sum_ranks()
```

```python
import numpy as np


@flock.distribute(workers=4)
async def sum_array():
    return await flock.all_gather(np.array([1, 2, 3]))
    

sum_array()
```
